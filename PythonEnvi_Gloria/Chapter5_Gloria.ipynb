Investigate on the difference between is and == given the following variables.

```python
a = 2
b = 2.0
c = a

The operator == compares the values of two variables, while is compares their identity in memory.

Given a = 2, b = 2.0, and c = a:

a == b is True because they have the same value.

a is b is False because they are different objects (int vs float).

a == c is True and a is c is True because c refers to the same object as a.

Therefore, == checks value equality, while is checks whether two variables point to the same object.
```

Challenge 1

Write a program which repeatedly reads numbers until the user enters "done". Once "done" is entered, print out the total, count, and average of the numbers. If the user enters anything other than a number, detect their mistake using try and except and print an error message and skip to the next number.

Enter a number: 4

Enter a number: 5

Enter a number: bad data

Invalid input

Enter a number: 7

Enter a number: done

16 3 5.333333333333333

```python
total = 0
count = 0

while True:
    user_input = input("Enter a number: ")

    if user_input == "done":
        break

    try:
        number = float(user_input)
    except ValueError:
        print("Invalid input")
        continue

    total += number
    count += 1

if count > 0:
    average = total / count
    print(int(total), count, average)
else:
    print("No valid numbers entered")
```

Challenge 2

Write another program that prompts for a list of numbers as above and at the end prints out both the maximum and minimum of the numbers instead of the average.

```python
largest = None
smallest = None

while True:
    user_input = input("Enter a number: ")

    if user_input == "done":
        break

    try:
        number = float(user_input)
    except ValueError:
        print("Invalid input")
        continue

    if largest is None or number > largest:
        largest = number
    if smallest is None or number < smallest:
        smallest = number

if largest is not None and smallest is not None:
    print("Maximum:", largest)
    print("Minimum:", smallest)
else:
    print("No valid numbers entered")
```
