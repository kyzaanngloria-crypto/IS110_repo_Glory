## A. Problem Definition

### A.1 Brief Description
The **Breast Cancer Wisconsin (Diagnostic)** dataset contains measurements computed from digitized images of fine-needle aspirates (FNA) of breast masses. Each row describes the characteristics of the cell nuclei present in one image.

- **569 instances**, **30 numeric features**, **1 binary target**
- Features are 10 nucleus measurements (radius, texture, perimeter, area, smoothness, compactness, concavity, concave points, symmetry, fractal dimension), each reported as the **mean**, the **standard error** and the **worst** (largest) value.
- Target: `0` = malignant, `1` = benign.

### A.2 Problem Statement
✏️ **Exercise A.1.** In 3–4 sentences, write a problem statement in your own words. Include: (a) the prediction task, (b) the input data, (c) the type of learning problem, (d) which kind of error is more costly and why, and (e) how you will judge success.

> **Your answer:

The task is to predict whether a breast mass is malignant or benign from 30 numeric measurements (mean, standard error and worst value of 10 nucleus features) computed from digitized fine-needle aspirate images. This is a supervised, binary classification problem trained on labeled examples (569 instances). A false negative (calling a malignant tumor benign) is more costly than a false positive, because a missed cancer delays treatment while a false alarm leads only to extra testing. I will judge success mainly by recall on the malignant class, alongside accuracy, precision and the F1 score, using cross-validation on training data and a final held-out test set.

## D. Evaluate Algorithm

We use **stratified 10-fold cross-validation** on the training set as our test harness. Scaling is placed inside a `Pipeline` so it is re-fitted in every fold: this avoids leakage.

**Exercise D.1.** Define the six candidate models: Logistic Regression, LDA, KNN, CART, Gaussian Naive Bayes and SVM.

```python
seed = 7
kfold = StratifiedKFold(n_splits=10, shuffle=True, random_state=seed)

# TODO: dictionary named models with keys LR, LDA, KNN, CART, NB, SVM
models = {
    "LR":   LogisticRegression(max_iter=1000),
    "LDA":  LinearDiscriminantAnalysis(),
    "KNN":  KNeighborsClassifier(),
    "CART": DecisionTreeClassifier(random_state=seed),
    "NB":   GaussianNB(),
    "SVM":  SVC(),
}

def spot_check(models, X_tr, scale):
    """Cross-validated accuracy for every model (optionally scaled inside the pipeline)."""
    results = {}
    for name, model in models.items():
        steps = ([("scale", StandardScaler())] if scale else []) + [("model", model)]
        results[name] = cross_val_score(Pipeline(steps), X_tr, y_train,
                                        cv=kfold, scoring="accuracy")
    return results
```

**Exercise D.2.** Run the spot-check on all features, once **raw** and once **scaled**, and tabulate the mean accuracies.

```python
# TODO: call spot_check twice and build a DataFrame called summary with columns raw / scaled
res_raw    = spot_check(models, X_train, scale=False)
res_scaled = spot_check(models, X_train, scale=True)

summary = pd.DataFrame({
    "raw":    {k: v.mean() for k, v in res_raw.items()},
    "scaled": {k: v.mean() for k, v in res_scaled.items()},
}).round(4)
display(summary)
```

✏️ **Question D.3.** Which models benefit from scaling, and why?

> **Your answer:**

KNN, SVM, logistic regression (and to a lesser degree LDA) benefit from scaling. KNN and SVM depend on distances, so features with large ranges such as area dominate unless scaled, and logistic regression’s optimizer converges better and its regularization treats features fairly on a common scale. Decision trees and Naive Bayes are essentially unaffected, because tree splits depend only on the ordering of values and Naive Bayes models each feature separately.

**Exercise D.4.** Repeat with the `selected` features only (scaled). Does feature selection help?

```python
# TODO: spot_check on X_train[selected] and compare against the all-features scaled results
res_sel = spot_check(models, X_train[selected], scale=True)
compare = pd.DataFrame({
    "all features (scaled)": {k: v.mean() for k, v in res_scaled.items()},
    "selected (scaled)":     {k: v.mean() for k, v in res_sel.items()},
}).round(4)
display(compare)
```

**Exercise D.5.** Compare the distributions of the scaled models with a box plot.

```python
# TODO: box plot of the 10 fold scores per model
plt.figure(figsize=(8, 5))
plt.boxplot(list(res_scaled.values()))
plt.xticks(range(1, len(res_scaled) + 1), res_scaled.keys())
plt.ylabel("10-fold accuracy")
plt.title("Scaled models: CV score distributions")
plt.show()
```

**Exercise D.6.** In a medical setting accuracy is not enough. For the best scaled model, cross-validate **accuracy, ROC-AUC and recall for the malignant class** (`pos_label=0`).

```python
# TODO: cross_validate with a scoring dict including make_scorer(recall_score, pos_label=0)
best_name = summary["scaled"].idxmax()
best_pipe = Pipeline([("scale", StandardScaler()), ("model", models[best_name])])

scoring = {
    "accuracy": "accuracy",
    "roc_auc": "roc_auc",
    "recall_malignant": make_scorer(recall_score, pos_label=0),
}
cvres = cross_validate(best_pipe, X_train, y_train, cv=kfold, scoring=scoring)
print("Best model:", best_name)
for k in scoring:
    print(f"{k}: {cvres['test_' + k].mean():.4f} ± {cvres['test_' + k].std():.4f}")
```

## E. Improve Accuracy

### E.1 Algorithm Tuning
**Exercise E.1.** Tune the SVM (`C`, `gamma`, `kernel`) with `GridSearchCV` inside a scaling pipeline.

```python
# TODO: grid search over C=[0.1,1,10,100], gamma=["scale",0.001,0.01,0.1], kernel=["rbf","linear"]
svm_pipe = Pipeline([("scale", StandardScaler()), ("model", SVC())])
svm_grid = {
    "model__C": [0.1, 1, 10, 100],
    "model__gamma": ["scale", 0.001, 0.01, 0.1],
    "model__kernel": ["rbf", "linear"],
}
gs_svm = GridSearchCV(svm_pipe, svm_grid, cv=kfold, scoring="accuracy", n_jobs=-1)
gs_svm.fit(X_train, y_train)
print(gs_svm.best_params_, round(gs_svm.best_score_, 4))
best_svm = gs_svm.best_estimator_
```

**Exercise E.2.** Tune KNN (`n_neighbors`, `weights`) the same way.

```python
# TODO: grid over n_neighbors=[3,5,7,9,11,15,21] and weights=["uniform","distance"]
knn_pipe = Pipeline([("scale", StandardScaler()), ("model", KNeighborsClassifier())])
knn_grid = {
    "model__n_neighbors": [3, 5, 7, 9, 11, 15, 21],
    "model__weights": ["uniform", "distance"],
}
gs_knn = GridSearchCV(knn_pipe, knn_grid, cv=kfold, scoring="accuracy", n_jobs=-1)
gs_knn.fit(X_train, y_train)
print(gs_knn.best_params_, round(gs_knn.best_score_, 4))
best_knn = gs_knn.best_estimator_
```

### E.2 Ensembles
**Exercise E.3.** Spot-check four ensemble methods (Random Forest, Extra Trees, Gradient Boosting, AdaBoost) and one voting ensemble of the tuned SVM, tuned KNN and Random Forest.

```python
# TODO: dictionary ensembles, score with spot_check(scale=False), then a hard VotingClassifier
ensembles = {
    "RF":  RandomForestClassifier(n_estimators=200, random_state=seed),
    "ET":  ExtraTreesClassifier(n_estimators=200, random_state=seed),
    "GBM": GradientBoostingClassifier(random_state=seed),
    "AB":  AdaBoostClassifier(random_state=seed),
}
res_ens = spot_check(ensembles, X_train, scale=False)

voting = VotingClassifier(
    estimators=[("svm", best_svm), ("knn", best_knn),
                ("rf", RandomForestClassifier(n_estimators=200, random_state=seed))],
    voting="hard")
res_vote = cross_val_score(voting, X_train, y_train, cv=kfold, scoring="accuracy")
print({k: round(v.mean(), 4) for k, v in res_ens.items()}, round(res_vote.mean(), 4))
```

**Exercise E.4.** Build a leaderboard (Series sorted by mean CV accuracy) containing both tuned models, the four ensembles and the voting ensemble.

```python
# TODO: leaderboard Series
leaderboard = pd.Series({
    "Tuned SVM": cross_val_score(best_svm, X_train, y_train, cv=kfold).mean(),
    "Tuned KNN": cross_val_score(best_knn, X_train, y_train, cv=kfold).mean(),
    **{k: v.mean() for k, v in res_ens.items()},
    "Voting": res_vote.mean(),
}).sort_values(ascending=False).round(4)
print(leaderboard)
```

### E.3 Finalize the Model
**Exercise E.5.** Fit the winner on the full training set and, for the first and only time, evaluate it on the held-out **test set**. Report accuracy, the classification report and a confusion matrix.

```python
# TODO: pick the leaderboard winner, fit on train, predict test, print metrics, plot confusion matrix
final_models = {"Tuned SVM": best_svm, "Tuned KNN": best_knn,
                **ensembles, "Voting": voting}
winner = leaderboard.index[0]
final = final_models[winner].fit(X_train, y_train)
y_pred = final.predict(X_test)

print("Winner:", winner)
print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print(classification_report(y_test, y_pred, target_names=["malignant", "benign"]))
ConfusionMatrixDisplay.from_predictions(
    y_test, y_pred, display_labels=["malignant", "benign"])
plt.show()
```

**Exercise E.6.** Save the model with `joblib`, reload it and confirm it reproduces predictions.

```python
# TODO: joblib.dump, joblib.load, compare first 5 predictions with y_test
import joblib
joblib.dump(final, "best_model.joblib")
loaded = joblib.load("best_model.joblib")
print("Predictions:", loaded.predict(X_test.iloc[:5]).tolist())
print("True labels:", y_test.iloc[:5].tolist())
print("Identical to original:", np.array_equal(loaded.predict(X_test), y_pred))
```

1. Leakage. When features are selected using the whole dataset, the selection step has already seen the labels of every row, including those that later act as validation folds. The chosen features are therefore tuned to the validation data, and CV scores come out optimistically high. The fix is to put SelectKBest (and the scaler) inside a Pipeline, so that cross-validation refits selection on each training fold only and evaluates on the untouched fold.

2. Metric choice. I would report recall (sensitivity) for the malignant class, together with precision and ROC-AUC. Accuracy alone is insufficient because the classes are imbalanced and because the two errors are not equally costly: a missed malignancy (false negative) can delay life-saving treatment, whereas a false positive usually leads only to follow-up testing.

3. Model choice. A 0.3% gap is likely within fold-to-fold noise, so I would compare the standard deviation across folds, not just the means. If the models are statistically indistinguishable, I would prefer the one that is more interpretable (e.g. logistic regression, which a clinician can inspect), cheaper to train and serve, and more stable across folds.

4. Generalization. With a much smaller dataset, CV estimates become noisier, complex models overfit more, and simple regularized models are safer. With a much more imbalanced dataset, accuracy becomes misleading, so I would rely on recall, precision-recall AUC and F1, use stratified splits, and consider class weights or resampling.

5. Extension. For a new dataset loaded from ARFF (scipy.io.arff.loadarff), the changes would be the loading cell (and decoding byte strings and encoding the target), the class names and pos_label for the positive class, handling any categorical features or missing values in Part C, and re-choosing the evaluation metric to fit the new problem’s costs. The overall workflow stays the same.
